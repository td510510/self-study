## 1. List — dãy có thứ tự

````python
nums = [3, 1, 4, 1, 5]

print(nums[0], nums[-1])          # đầu, cuối
print(nums[1:3])                  # cắt lát: từ 1 đến TRƯỚC 3
print(nums[:2], nums[2:])
print(nums[::-1])                 # đảo ngược
print(nums[::2])                  # lấy cách một
````

````python
nums = [3, 1, 4]
nums.append(9)     ; print("append :", nums)
nums.insert(0, 0)  ; print("insert :", nums)
nums.remove(1)     ; print("remove :", nums)
cuoi = nums.pop()  ; print("pop    :", nums, "-> lấy ra", cuoi)
nums.sort()        ; print("sort   :", nums)
print("sorted giảm dần:", sorted(nums, reverse=True))
print("list gốc không đổi:", nums)
````

> `sort()` sửa list gốc và trả về `None`. `sorted()` trả về list mới. Nhầm hai cái này là lỗi rất phổ biến:
> ```python
> nums = nums.sort()   # ❌ nums thành None!
> ```

## 2. Dict — tra cứu theo khoá

````python
nguoi = {"ten": "An", "tuoi": 25}

print(nguoi["ten"])
print(nguoi.get("email"))                # None — không lỗi
print(nguoi.get("email", "chưa có"))     # giá trị mặc định

nguoi["email"] = "an@x.com"              # thêm
del nguoi["tuoi"]                        # xoá
print(nguoi)
print("ten" in nguoi)
````

````python
# @loi-co-y  <- ô này CỐ Ý lỗi, dùng để bạn xem thông báo lỗi
# Truy cập khoá không tồn tại bằng [] -> KeyError
nguoi["so_dien_thoai"]
````

> Dùng `.get()` khi khoá **có thể** không tồn tại. Dùng `[]` khi khoá **bắt buộc** phải có — lúc đó bạn *muốn* nó báo lỗi ngay.

````python
# Đếm bằng dict — mẫu rất hay dùng
van_ban = "meo an ca meo ngu"
dem = {}
for tu in van_ban.split():
    dem[tu] = dem.get(tu, 0) + 1
print(dem)

# Thư viện chuẩn đã có sẵn
from collections import Counter
print(Counter(van_ban.split()))
print(Counter(van_ban.split()).most_common(2))
````

## 3. Set — không trùng, tra cực nhanh

````python
a = {1, 2, 3}
b = {3, 4, 5}

print("giao     a & b =", a & b)
print("hợp      a | b =", a | b)
print("hiệu     a - b =", a - b)
print("đối xứng a ^ b =", a ^ b)

print("\nloại trùng:", set([1, 1, 2, 2, 3]))
````

### ⚡ Vì sao set nhanh hơn list — đo thử

````python
import time

n = 50_000
ds_list = list(range(n))
ds_set = set(ds_list)
can_tim = list(range(n - 1000, n))     # 1000 phần tử ở cuối

t = time.perf_counter()
for x in can_tim:
    _ = x in ds_list
t_list = time.perf_counter() - t

t = time.perf_counter()
for x in can_tim:
    _ = x in ds_set
t_set = time.perf_counter() - t

print(f"list : {t_list*1000:8.2f} ms")
print(f"set  : {t_set*1000:8.2f} ms")
print(f"set nhanh hơn khoảng {t_list/max(t_set, 1e-9):.0f} lần")
````

**Vì sao?**
- `in` trên **list** phải quét từng phần tử → O(n)
- `in` trên **set** hoặc **dict** dùng bảng băm, tới thẳng chỗ cần → O(1)

Với dữ liệu 100.000 dòng, khác biệt là vài phút so với vài phần nghìn giây. Đây là loại quyết định phân biệt "biết cú pháp" với "kỹ sư".

## 4. Tuple — nhóm cố định

````python
toa_do = (10.5, 20.3)
x, y = toa_do                    # giải nén
print(x, y)

# Tuple bất biến -> dùng được làm khoá dict
kho = {(10, 20): "Hà Nội", (30, 40): "HCM"}
print(kho[(10, 20)])
````

````python
# @loi-co-y  <- ô này CỐ Ý lỗi, dùng để bạn xem thông báo lỗi
# List KHÔNG làm khoá dict được — chạy để xem lỗi
{[1, 2]: "test"}
````

> `TypeError: unhashable type: 'list'` — chỉ đối tượng **bất biến** mới băm được, nên chỉ chúng mới làm khoá dict / phần tử set được.

## 5. Comprehension

````python
nums = [1, -2, 3, -4, 5]

print([x * 2 for x in nums])                          # biến đổi
print([x for x in nums if x > 0])                     # lọc
print([x * 2 for x in nums if x > 0])                 # cả hai
print([("dương" if x > 0 else "âm") for x in nums])   # if/else trong biểu thức

d = {"a": 1, "b": 0, "c": 3}
print({k: v for k, v in d.items() if v})              # dict comprehension
print({x % 3 for x in nums})                          # set comprehension
````

````python
# Lồng hai vòng — làm phẳng list của list
long_nhau = [[1, 2], [3], [4, 5]]
print([x for con in long_nhau for x in con])

# Đọc theo đúng thứ tự vòng for, y hệt vòng lặp thường:
kq = []
for con in long_nhau:
    for x in con:
        kq.append(x)
print(kq)
````

## 6. Bảng chọn cấu trúc dữ liệu

| Câu hỏi | Dùng |
|---|---|
| Cần giữ thứ tự, cho phép trùng | `list` |
| Tra cứu theo khoá, tốc độ quan trọng | `dict` |
| Chỉ cần biết "có hay không", loại trùng | `set` |
| Nhóm giá trị cố định, làm khoá dict | `tuple` |
