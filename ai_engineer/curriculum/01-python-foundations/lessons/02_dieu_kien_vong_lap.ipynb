## 1. if / elif / else — thứ tự quyết định kết quả

````python
diem = 9.5

# ĐÚNG: xét từ cao xuống thấp
if diem >= 9:
    print("Giỏi")
elif diem >= 6.5:
    print("Trung bình")
else:
    print("Yếu")
````

````python
# SAI: thứ tự ngược -> điểm 9.5 rơi vào nhánh đầu tiên khớp
diem = 9.5
if diem >= 6.5:
    print("Trung bình")     # <- chạy nhánh này, dừng luôn
elif diem >= 9:
    print("Giỏi")           # <- KHÔNG BAO GIỜ chạy tới
````

> Đây là bug **thầm lặng**: code chạy bình thường, không báo lỗi, chỉ cho kết quả sai. Loại bug khó chịu nhất.

## 2. Giá trị falsy

````python
for v in [False, None, 0, 0.0, "", [], {}, set(), "0", [0], " "]:
    print(f"{repr(v):<8} -> {'truthy' if v else 'falsy'}")
````

Chú ý ba dòng cuối: chuỗi `"0"`, list `[0]` và chuỗi khoảng trắng `" "` đều **truthy** — chúng không rỗng.

````python
ds = []
if not ds:                       # cách viết Python-ish
    print("Danh sách rỗng")

# ⚠️ NHƯNG cẩn thận khi 0 là giá trị hợp lệ
so_luong = 0
if not so_luong:
    print("Bẫy: 0 cũng bị coi là 'không có giá trị'")
if so_luong is None:
    print("Cách này mới đúng khi 0 là hợp lệ")
````

## 3. Vòng lặp for

````python
print("range:", list(range(5)), list(range(2, 10, 2)), list(range(5, 0, -1)))
````

````python
ten = ["An", "Bình", "Cường"]
diem = [8, 9, 7]

print("--- duyệt phần tử ---")
for t in ten:
    print(t)

print("\n--- kèm chỉ số ---")
for i, t in enumerate(ten, start=1):
    print(f"{i}. {t}")

print("\n--- song song hai list ---")
for t, d in zip(ten, diem):
    print(f"{t}: {d}")

print("\n--- duyệt dict ---")
for k, v in dict(zip(ten, diem)).items():
    print(f"{k} -> {v}")
````

> Nếu bạn đang viết `for i in range(len(ds)):` thì gần như luôn có cách hay hơn — dùng `enumerate` hoặc duyệt trực tiếp.

## 4. break / continue

````python
for x in [1, -2, 3, 150, 4]:
    if x < 0:
        print(f"  bỏ qua {x}")
        continue
    if x > 100:
        print(f"  gặp {x}, dừng hẳn")
        break
    print(f"  xử lý {x}")
````

## 5. ⚠️ Không bao giờ sửa list khi đang duyệt nó

````python
nums = [1, -2, -3, 4]

for x in nums:          # SAI
    if x < 0:
        nums.remove(x)

print(nums)             # -3 vẫn còn — bị bỏ sót!
````

**Vì sao?** Khi xoá `-2` (vị trí 1), các phần tử phía sau dồn lên một bậc, nhưng vòng lặp vẫn nhảy sang vị trí 2 → `-3` bị nhảy qua.

**Cách đúng:** tạo list mới.

````python
nums = [1, -2, -3, 4]
nums = [x for x in nums if x >= 0]
print(nums)
````

## 6. while — khi không biết trước số lần lặp

````python
so_du = 1000
nam = 0

while so_du < 2000:
    so_du *= 1.1
    nam += 1

print(f"Sau {nam} năm, số dư = {so_du:,.0f}")
````

> ⚠️ Vòng `while` phải có điều kiện chắc chắn sẽ sai đi, nếu không nó chạy mãi mãi. Bấm `Ctrl+C` (hoặc nút ⏹ trong notebook) để dừng.
