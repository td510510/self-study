## 1. Biến là cái NHÃN, không phải cái hộp

Đây là mô hình tinh thần quan trọng nhất của bài này.

````python
a = [1, 2, 3]
b = a            # b dán lên CHÍNH list đó, không phải bản sao

b.append(4)
print("a =", a)
print("b =", b)
print("a và b có phải cùng một đối tượng?", a is b)
````

````python
# Muốn độc lập thì phải copy
a = [1, 2, 3]
b = a.copy()

b.append(4)
print("a =", a)
print("b =", b)
print("cùng đối tượng?", a is b)
````

> 🔬 **Thử nghiệm:** đổi `a.copy()` thành `list(a)` rồi `a[:]`. Cả ba đều tạo bản sao.

### Với kiểu bất biến thì không có vấn đề này

````python
x = 5
y = x
y = 10
print("x =", x, "| y =", y)     # x không đổi

s = "hello"
t = s
t = t.upper()                    # tạo chuỗi MỚI, không sửa chuỗi cũ
print("s =", s, "| t =", t)
````

| Sửa được (mutable) | Bất biến (immutable) |
|---|---|
| `list`, `dict`, `set` | `int`, `float`, `str`, `bool`, `tuple` |

Bất biến nghĩa là mọi thao tác đều tạo ra giá trị **mới**, không sửa cái cũ.

## 2. Bốn kiểu dữ liệu cơ bản

````python
gia_tri = [42, 3.14, "xin chào", True, None, [1, 2], {"a": 1}, (1, 2), {1, 2}]

for v in gia_tri:
    print(f"{repr(v):<15} -> {type(v).__name__}")
````

### Ép kiểu

````python
print(int("42"), type(int("42")))
print(float("3.14"))
print(str(42) + " tuổi")
print(int(3.99))          # CẮT phần thập phân, không làm tròn!
print(round(3.99))        # muốn làm tròn thì dùng round()
print(bool(0), bool(""), bool([]), bool("a"))
````

## 3. Chuỗi — thứ bạn xử lý nhiều nhất khi làm AI

````python
s = "  Nguyen Van Nam  "

print(repr(s.strip()))         # bỏ khoảng trắng 2 đầu
print(s.strip().lower())
print(s.strip().upper())
print(s.strip().split())       # tách thành list
print(s.strip().replace("Nam", "An"))
print(len(s), "->", len(s.strip()))
````

````python
# split() không tham số vs split(" ") — khác biệt QUAN TRỌNG
print("a  b".split())        # ['a', 'b']       — gộp khoảng trắng thừa
print("a  b".split(" "))     # ['a', '', 'b']   — có phần tử rỗng!
````

### f-string — cách nối chuỗi nên dùng

````python
ten, tuoi, diem = "Nam", 25, 8.567

print(f"{ten} - {tuoi} tuổi")
print(f"Điểm: {diem:.1f}")               # 1 chữ số thập phân
print(f"Tiền: {1234567:,.0f} VND")       # dấu phân cách hàng nghìn
print(f"Tỷ lệ: {0.8567:.1%}")            # tự nhân 100 và thêm %
print(f"{ten:>10}|")                     # căn phải, độ rộng 10
print(f"{ten:<10}|")                     # căn trái
print(f"{ten:^10}|")                     # căn giữa
print(f"{diem=}")                        # in cả tên biến — rất tiện khi debug
````

## 4. Toán tử

````python
a, b = 7, 3
for ky_hieu, ket_qua in [
    ("a + b", a + b), ("a - b", a - b), ("a * b", a * b),
    ("a / b", a / b), ("a // b", a // b), ("a % b", a % b), ("a ** b", a ** b),
]:
    print(f"{ky_hieu:<8} = {ket_qua}")
````

> `//` và `%` đi với nhau rất thường xuyên khi chia đơn vị (phút→giờ, giây→phút).
> `7 // 3 = 2` giờ và `7 % 3 = 1` phút dư.

````python
# Số thực KHÔNG chính xác tuyệt đối — điều mọi lập trình viên phải biết
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)          # False!

# Cách so sánh đúng
print(abs((0.1 + 0.2) - 0.3) < 1e-9)
````

**Vì sao?** Máy tính lưu số thực ở hệ nhị phân; `0.1` không biểu diễn chính xác được, giống như `1/3` không viết hết được ở hệ thập phân. Đây là lý do **không bao giờ dùng `float` cho tiền tệ** trong hệ thống thật (dùng `int` đơn vị đồng, hoặc `Decimal`).
