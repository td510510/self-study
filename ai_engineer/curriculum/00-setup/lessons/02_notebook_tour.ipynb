## 3. ⚠️ Cạm bẫy: thứ tự chạy ≠ thứ tự hiển thị

Đây là bài học quan trọng nhất của notebook. Làm thí nghiệm sau **đúng từng bước**:

1. Chạy ô A bên dưới
2. Chạy ô B → thấy `10`
3. Quay lên **sửa** ô A thành `x = 99` nhưng **KHÔNG chạy lại nó**
4. Chạy lại ô B → nó vẫn in `10`, không phải `99`!

````python
# Ô A
x = 10
````

````python
# Ô B
print(f"x hiện tại = {x}")
````

**Vì sao?** Notebook không đọc chữ trên màn hình — nó chỉ nhớ những gì bạn **đã thật sự chạy**. Sửa code mà không chạy lại thì bộ nhớ vẫn giữ giá trị cũ.

Con số trong `[ ]` bên trái mỗi ô cho biết **thứ tự chạy thật**. Nếu số của ô A nhỏ hơn ô B, tức là A chạy trước — nhưng nếu bạn sửa A sau đó, số không đổi mà nội dung đã khác.

### 🛡 Thói quen bắt buộc

Trước khi tin vào kết quả của bất kỳ notebook nào:

> **Kernel → Restart Kernel and Run All Cells**

Nếu chạy lại từ đầu vẫn ra kết quả đó thì mới là thật. Không làm bước này là nguyên nhân của vô số giờ debug vô ích — kể cả với người đã đi làm nhiều năm.

## 4. Markdown — ghi chú ngay trong notebook

Ô bạn đang đọc là ô **Markdown**, không phải code. Đây là điều làm notebook tuyệt vời để học và để báo cáo: giải thích và code nằm cạnh nhau.

Cú pháp hay dùng:

| Viết | Ra |
|---|---|
| `**đậm**` | **đậm** |
| `*nghiêng*` | *nghiêng* |
| `` `code` `` | `code` |
| `# Tiêu đề` | Tiêu đề lớn |
| `- mục` | danh sách |

**Đổi loại ô:** bấm `Esc` rồi `M` (thành Markdown) hoặc `Y` (thành Code).

## 5. Phím tắt đáng thuộc

Bấm `Esc` trước (thoát chế độ gõ), rồi:

| Phím | Tác dụng |
|---|---|
| `A` | Thêm ô phía **trên** (Above) |
| `B` | Thêm ô phía **dưới** (Below) |
| `D` `D` | Xoá ô (bấm D hai lần) |
| `M` | Chuyển thành ô Markdown |
| `Y` | Chuyển thành ô Code |
| `Z` | Hoàn tác xoá ô |
| `Shift + Enter` | Chạy ô và xuống dưới |

Bấm `Enter` để quay lại chế độ gõ trong ô.

## 7. Khi nào dùng notebook, khi nào dùng file .py

| Việc | Dùng gì |
|---|---|
| Khám phá dữ liệu mới, vẽ biểu đồ | 📓 Notebook |
| Thử nhanh một ý tưởng | 📓 Notebook |
| Dạy / báo cáo có kèm giải thích | 📓 Notebook |
| Viết hàm dùng lại nhiều nơi | 📄 `.py` |
| Viết test | 📄 `.py` |
| Code chạy thật, deploy | 📄 `.py` |

**Quy trình thực tế trong nghề:** khám phá trong notebook → khi một đoạn code đã ổn định, **chuyển nó thành hàm trong file `.py`** → notebook chỉ còn gọi hàm đó.
